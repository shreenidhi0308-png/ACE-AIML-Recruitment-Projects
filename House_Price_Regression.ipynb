```python
# Loading the training and test datasets

train = pd.read_csv('/kaggle/input/datasets/shreenidhi1503/part-1/train.csv')
test = pd.read_csv('/kaggle/input/datasets/shreenidhi1503/part-1/test.csv')

hpd = train
```

```python
# Separating the features and target variable
X = hpd.drop(['SalePrice'], axis=1)
Y = hpd['SalePrice']
X = pd.get_dummies(X)
X = X.astype(float)
```

```python
# Viewing the features and target
print(X)
print(Y)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)
```

```python
import numpy as np
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from sklearn.metrics import mean_squared_log_error

# Train using log prices
Y_train_log = np.log1p(Y_train)

# XGBoost
xgb = XGBRegressor(
    n_estimators=3000,
    learning_rate=0.015,
    max_depth=4,
    min_child_weight=2,
    subsample=0.85,
    colsample_bytree=0.85,
    reg_alpha=0.03,
    reg_lambda=2.0,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

# LightGBM
lgb = LGBMRegressor(
    n_estimators=3000,
    learning_rate=0.015,
    num_leaves=31,
    max_depth=5,
    min_child_samples=15,
    subsample=0.85,
    colsample_bytree=0.85,
    reg_alpha=0.03,
    reg_lambda=1.5,
    random_state=42,
    verbosity=-1,
    n_jobs=-1
)

# CatBoost
cat = CatBoostRegressor(
    iterations=2500,
    learning_rate=0.025,
    depth=6,
    l2_leaf_reg=5,
    loss_function="RMSE",
    random_seed=42,
    verbose=False
)

# Train
xgb.fit(X_train, Y_train_log)
lgb.fit(X_train, Y_train_log)
cat.fit(X_train, Y_train_log)

# Validation predictions
p_xgb = xgb.predict(X_test)
p_lgb = lgb.predict(X_test)
p_cat = cat.predict(X_test)

# Ensemble
pred_log = (
    0.40 * p_xgb +
    0.35 * p_lgb +
    0.25 * p_cat
)

# Convert back to price
pred = np.expm1(pred_log)
pred = np.maximum(pred, 0)

# Validation MSLE
msle = mean_squared_log_error(Y_test, pred)

print("Validation MSLE:", msle)
```

```python
Y_log = np.log1p(Y)

xgb_final = XGBRegressor(
    n_estimators=3000,
    learning_rate=0.015,
    max_depth=4,
    min_child_weight=2,
    subsample=0.85,
    colsample_bytree=0.85,
    reg_alpha=0.03,
    reg_lambda=2.0,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

lgb_final = LGBMRegressor(
    n_estimators=3000,
    learning_rate=0.015,
    num_leaves=31,
    max_depth=5,
    min_child_samples=15,
    subsample=0.85,
    colsample_bytree=0.85,
    reg_alpha=0.03,
    reg_lambda=1.5,
    random_state=42,
    verbosity=-1,
    n_jobs=-1
)

cat_final = CatBoostRegressor(
    iterations=2500,
    learning_rate=0.025,
    depth=6,
    l2_leaf_reg=5,
    loss_function="RMSE",
    random_seed=42,
    verbose=False
)

xgb_final.fit(X, Y_log)
lgb_final.fit(X, Y_log)
cat_final.fit(X, Y_log)

print("Final models trained.")
```

```python
# Prepare test features
competition_X = test.drop(columns=['PID'], errors='ignore')

# One-hot encode test
competition_X = pd.get_dummies(competition_X)

# Match training columns exactly
competition_X = competition_X.reindex(
    columns=X.columns,
    fill_value=0
)

# Predict
pred_xgb = xgb_final.predict(competition_X)
pred_lgb = lgb_final.predict(competition_X)
pred_cat = cat_final.predict(competition_X)

# Ensemble in log space
final_log_prediction = (
    0.40 * pred_xgb +
    0.35 * pred_lgb +
    0.25 * pred_cat
)

# Convert log predictions back to prices
final_prediction = np.expm1(final_log_prediction)
final_prediction = np.maximum(final_prediction, 0)

# Submission
submission = pd.DataFrame({
    'PID': test['PID'],
    'SalePrice': final_prediction
})

submission.to_csv(
    '/kaggle/working/submission.csv',
    index=False
)

print(submission.head())
print("Submission shape:", submission.shape)
```
